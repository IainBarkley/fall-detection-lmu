```python
from pathlib import Path

tflite_path = Path('~/repos/summerschool2023/projects/fall-detection/fall_detection_data/models/quantized/micro_cnn_fold_5_int8.tflite').expanduser()

data = tflite_path.read_bytes()

header = f"""// Micro-CNN Fall Detection — Float16, Fold 5
// Size: {len(data)} bytes ({len(data)/1024:.1f} KB)
// Accuracy: 95.01% | Fall_Init Recall: 98.79%
// Target: Arduino Nano 33 BLE Sense (nRF52840)

#ifndef FALLNET_MODEL_H
#define FALLNET_MODEL_H

const unsigned int fallnet_model_len = {len(data)};
alignas(8) const unsigned char fallnet_model[] = {{
"""

for i in range(0, len(data), 12):
    chunk = data[i:i+12]
    header += '  ' + ', '.join(f'0x{b:02x}' for b in chunk) + ',\n'

header += "};\n\n#endif // FALLNET_MODEL_H\n"

out_path = tflite_path.parent / 'fallnet_model.h'
out_path.write_text(header)
print(f"✅ {out_path}  ({len(data)/1024:.1f} KB)")
```
